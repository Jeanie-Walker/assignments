```python
from datetime import date, datetime

current = date.today().isoformat()
with open("today.txt", "wt") as file:
    file.write(current)
```

```python
with open("today.txt", "rt") as file:
    today = file.read()
```

```python
format = "%Y-%m-%d"
parsed = datetime.strptime(today, format)
print(parsed)
print(parsed.date())
```

```python
import multiprocessing
import random
import time
from datetime import datetime

def wait():
    seconds = random.random()
    time.sleep(seconds)
    print(datetime.now())

if __name__ == '__main__':
    for num in range(3):
        process = multiprocessing.Process(target=wait)
        process.start()
```
